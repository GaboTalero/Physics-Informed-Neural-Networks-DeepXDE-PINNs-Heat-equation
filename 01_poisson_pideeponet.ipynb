# 01. Poisson PI-DeepONet

This notebook is a clean DeepXDE-native operator-learning example.

We learn the operator that maps a source term `f(x)` to the solution `u(x)` of the 1D Poisson equation:

`-u_xx = f(x)` on `[0, 1]`, with homogeneous Dirichlet boundary conditions.


```python
import numpy as np
import matplotlib.pyplot as plt

try:
    import deepxde as dde
except ImportError:
    dde = None
    print("DeepXDE is not installed. Install dependencies from requirements.txt before running this notebook.")
```

```python
if dde is None:
    raise ImportError("DeepXDE is required for this notebook.")

# Poisson equation: -u_xx = f

def equation(x, y, f):
    dy_xx = dde.grad.hessian(y, x)
    return -dy_xx - f

geom = dde.geometry.Interval(0, 1)


def boundary(_, on_boundary):
    return on_boundary


bc = dde.icbc.DirichletBC(geom, lambda _: 0, boundary)
pde = dde.data.PDE(geom, equation, bc, num_domain=100, num_boundary=2)

space = dde.data.PowerSeries(N=4)
evaluation_points = geom.uniform_points(10, boundary=True)

pde_op = dde.data.PDEOperatorCartesianProd(
    pde,
    space,
    evaluation_points,
    num_function=100,
)

net = dde.nn.DeepONetCartesianProd(
    [10, 32, 32],
    [1, 32, 32],
    activation="tanh",
    kernel_initializer="Glorot normal",
)

model = dde.Model(pde_op, net)
dde.optimizers.set_LBFGS_options(maxiter=1000)
model.compile("L-BFGS")
losshistory, train_state = model.train()
```

```python
features = space.random(3)
fx = space.eval_batch(features, evaluation_points)
x = geom.uniform_points(100, boundary=True)
y = model.predict((fx, x))

fig, axes = plt.subplots(2, 1, figsize=(8, 7), sharex=True)
axes[0].set_title("Poisson operator learning with PI-DeepONet")
for i in range(3):
    axes[0].plot(evaluation_points, fx[i], "--", label=f"f_{i+1}(x)")
axes[0].set_ylabel("source")
axes[0].legend()

for i in range(3):
    axes[1].plot(x, y[i], label=f"u_{i+1}(x)")
axes[1].set_xlabel("x")
axes[1].set_ylabel("solution")
axes[1].legend()
plt.tight_layout()
plt.show()
```
